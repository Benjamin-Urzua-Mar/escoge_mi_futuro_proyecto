**Pregunta de Investigación:** ¿Es posible identificar patrones en estudiantes universitarios que permitan orientar la elección entre una carrera técnica o una ingeniería?

Este notebook contiene el análisis exploratorio de datos (EDA), preprocesamiento y estructuración para alimentar la aplicación web interactiva en Streamlit.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-darkgrid')
print('Librerías cargadas exitosamente.')
```

### 1. Carga del Dataset Procesado
Cargamos los datos procesados desde `data_procesada.csv` generados a partir del registro oficial del DEMRE.

```python
df = pd.read_csv('data_procesada.csv')
print(f'Total de registros cargados: {len(df):,}')
df.head()
```

### 2. Distribución por Tipo de Carrera
Analizamos la proporción de postulaciones hacia Ingenierías vs. Carreras Técnicas y Otras.

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='tipo_carrera', palette='Set2')
plt.title('Distribución de Postulantes por Tipo de Carrera')
plt.xlabel('Tipo de Carrera')
plt.ylabel('Cantidad de Postulantes')
plt.show()
```

### 3. Comparativa del Puntaje NEM según Tipo de Carrera
Evaluamos la hipótesis de si los estudiantes que ingresan a Ingenierías poseen en promedio un puntaje NEM significativamente distinto.

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x='tipo_carrera', y='ptje_nem', palette='coolwarm')
plt.title('Comparación de Puntaje NEM por Tipo de Carrera')
plt.xlabel('Tipo de Carrera')
plt.ylabel('Puntaje NEM')
plt.show()
```

### 4. Conclusiones del EDA
1. **Dominio de Ingenierías:** En el registro de universidades del DEMRE, existe una mayor concentración de ofertas de carreras de Ingeniería.
2. **Comportamiento del NEM:** Los postulantes a Ingenierías muestran distribuciones sólidas de puntaje NEM y de matemáticas, lo que valida la inclusión de estos filtros en la aplicación Streamlit.
